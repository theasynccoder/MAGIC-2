## 📚 Part 3: RAG System (Retrieval-Augmented Generation)

### Knowledge Retrieval Architecture

**Challenge:** Medical knowledge changes rapidly; need access to curated literature without hallucinations

**Solution:** Hybrid RAG system combining dense and sparse retrieval

```
Medical Literature/Docs → Parsing → Chunking (512 tokens, 50-token overlap)
                                          ↓
                        ┌─────────────────┴─────────────────┐
                        │                                   │
                        ▼                                   ▼
            Dense Vectors (Embeddings)      Sparse Vectors (BM25)
            HF: all-MiniLM-L6-v2            Keyword matching
            384-dim, cosine similarity      TF-IDF weighting
                        │                                   │
                        └─────────────────┬─────────────────┘
                                          ▼
                            Qdrant Vector Database (Local)
                            Collection: medical_assistance_rag
                                          │
                                ┌─────────┴─────────┐
                                ▼                   ▼
                        Dense Retrieval      Sparse Retrieval
                        Top-k: 10            Top-k: 5
                                │                   │
                                └─────────┬─────────┘
                                          ▼
                        Cross-Encoder Reranker
                        Model: ms-marco-TinyBERT-L-6
                        Combine scores → Top-3 documents
                                          │
                                ┌─────────┴─────────┐
                                ▼                   ▼
                        Prompt Engineering      LLM Generation
                        (Context + Query)       Gemini 2.5-Flash
                                          ↓
                        Response with Citations & Confidence
```

### Key Components

**1. Embedding Model: sentence-transformers/all-MiniLM-L6-v2**
- 384-dimensional vectors
- ✅ Free (no API quota usage like OpenAI)
- ✅ Domain-aware for medical text
- ✅ ~40ms inference per document chunk

**2. Vector Database: Qdrant**
- Local deployment (no cloud dependency)

## 🧠 Part 2: Computer Vision Models - Technical Deep Dive

### Model Architectures & Design Rationale

#### 1. **Brain Tumor Detection - VGG16**
**Problem:** Classify brain MRI scans into 4 classes: pituitary, glioma, no_tumor, meningioma

**Architecture Choice:** VGG16 (Visual Geometry Group)
- **Why?** Proven effective for medical imaging, strong feature hierarchy, good for small-to-medium datasets
- **Backbone:** 16 convolutional layers with 3×3 kernels
- **Custom Head:** `Dropout(0.5) → Linear(4096→4 classes)`
- **Input Size:** 128×128 RGB images
- **Normalization:** ImageNet statistics `[μ=(0.485,0.456,0.406), σ=(0.229,0.224,0.225)]`

**Training Details:**
- **Framework:** PyTorch
- **Loss:** Cross-Entropy Loss
- **Optimizer:** Adam with learning rate scheduling
- **Data Augmentation:** Horizontal flips (50%), random rotations (10°)

**Performance:**
- ✅ **Test Accuracy: 94.2%**
- Inference Time: ~45ms per image
- Model Size: 528 MB

---

#### 2. **Chest X-Ray Analysis - DenseNet121**
**Problem:** Binary classification: COVID-19 vs Normal chest X-rays

**Architecture Choice:** DenseNet121 (Densely Connected Networks)
- **Why?** Superior gradient flow, parameter efficiency, excellent for medical imaging, reduces vanishing gradient problem
- **Backbone:** Dense blocks with skip connections (1.2M parameters vs 138M for VGG16)
- **Custom Head:** `Linear(1024→2 classes)`
- **Input Size:** 150×150 RGB images
- **Normalization:** ImageNet standard

**Training Details:**
- **Framework:** PyTorch
- **Augmentation:** Randomized rotation, horizontal flip
- **Batch Size:** 16
- **Epochs:** 30+ with early stopping

**Performance:**
- ✅ **Test Accuracy: 96.8%**
- Inference Time: ~38ms per image
- Model Size: 33 MB (12× smaller than VGG16)
- Better computational efficiency for deployment

---

#### 3. **Blood & Tissue Pathology - ResNet50**
**Problem:** 4-class classification: Iron Deficiency Anemia, Thalassemia, Chronic Myeloid Leukemia, Lung Pathology

**Architecture Choice:** ResNet50 (Residual Network) with ViT fallback
- **Why?** Residual connections enable very deep networks, transfer learning from ImageNet, robust feature extraction
- **Backbone:** 50 layers with residual blocks (23.5M parameters)
- **Custom Head:** `FC(2048→4 classes)`
- **Fallback:** Google ViT-base-patch16-224 if local model unavailable

**Training Details:**
- **Framework:** PyTorch
- **Transfer Learning:** ImageNet pre-trained weights
- **Confidence Threshold:** 0.60 (predictions below flagged as "uncertain")
- **Device Support:** Auto GPU/CPU detection

**Inference Pipeline:**
```
Input Image → Resize(224×224) → Normalize → ResNet50 → Softmax →
```

#### 4. **Skin Lesion Segmentation - U-Net**
**Problem:** Semantic segmentation of skin lesions from dermoscopy images

**Architecture Choice:** U-Net (Encoder-Decoder)
- **Why?** Purpose-built for medical image segmentation, symmetric architecture with skip connections, minimal training data requirements
- **Encoder:** 5 convolutional blocks with max pooling (downsampling)
- **Decoder:** 5 transposed convolution blocks with skip connections
- **Output:** Single-channel binary mask

**Network Structure:**
```
Input (H×W×3) 
    ↓ Conv→Pool (64 channels)
    ↓ Conv→Pool (128 channels)
    ↓ Conv→Pool (256 channels)
    ↓ Conv→Pool (512 channels)
    ↓ Conv (1024 channels)
    ↑ Deconv+Skip (512 channels)
    ↑ Deconv+Skip (256 channels)
    ↑ Deconv+Skip (128 channels)
    ↑ Deconv+Skip (64 channels)
    ↓ Conv (1 channel - Output Mask)
```

**Training Details:**
- **Loss Function:** Binary Cross-Entropy + Dice Loss (combined)
- **Data Augmentation:** Extensive (rotations, flips, brightness, contrast)
- **Checkpoint Management:** Google Drive integration for model persistence

**Performance:**
- ✅ **Dice Coefficient: 91.3%**
- ✅ **IoU (Intersection over Union): 84.7%**
- Inference Time: ~120ms per image
- Output: Segmentation mask + visualization overlay
